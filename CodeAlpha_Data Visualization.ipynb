```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Count books for each rating
rating_counts = df["Rating"].value_counts().sort_index()

# Calculate total number of books
total_books = len(df)

print("Number of Books for Each Rating:")
print(rating_counts)

print("\nTotal Number of Books:", total_books)

# Create bar chart
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

# Display count above each bar
ax.bar_label(ax.containers[0])

plt.title("Number of Books for Each Rating")
plt.xlabel("Rating (1–5)")
plt.ylabel("Number of Books")
plt.tight_layout()
plt.show()
```
